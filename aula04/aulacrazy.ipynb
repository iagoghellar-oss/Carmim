```python
def calcula_percentual (valor, percentual=100):
  quanto = valor * percentual / 100
  return quanto

print ("20% de 300 são: {:10.2f}" .format (calcula_percentual (300,20)))

print ("100% de 50 são: {:10.2f}" .format (calcula_percentual (50)))
```
